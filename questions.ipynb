```python
raw_incentives_dataset=spark.read.table('johndeere.cleansed_tables.raw_incentives_dataset')
display(raw_incentives_dataset)
```

Find the latest transaction per dealer using row_number().

```python
%sql
with ranked as (
select dealer_id,sale_id,tx_date,p_name,price,row_number()over(partition by dealer_id order by tx_date desc) as rn
from johndeere.cleansed_tables.raw_incentives_dataset
)
select * from ranked where rn=1
```
